## Projeto - Análise de Dados Meteorológicos

### Variáveis utilizadas

```python
temperaturas = []
meses_escaldantes = 0
maior_temperatura = -61
menor_temperatura = 51
mes_maior_temperatura = 0
mes_menor_temperatura = 0
```

### Laço para coletar dados para cada mês (1 a 12)

```python
for mes in range(1, 13):
    # Validando o mês
    while True:
        mes_usuario = int(input(f"Digite o número do mês ({mes}): "))
        if 1 <= mes_usuario <= 12:
            break
        else:
            print("Mês inválido! Digite um número entre 1 e 12.")
    
    # Validando a temperatura
    while True:
        temperatura = float(input(f"Digite a temperatura máxima do mês {mes_usuario}: "))
        if -60 <= temperatura <= 50:
            break
        else:
            print("Temperatura inválida! Digite um valor entre -60°C e 50°C.")
    
    # Armazenando a temperatura
    temperaturas.append(temperatura)

    # Contando os meses escaldantes (temperatura > 33°C)
    if temperatura > 33:
        meses_escaldantes += 1

    # Verificando o mês com maior temperatura
    if temperatura > maior_temperatura:
        maior_temperatura = temperatura
        mes_maior_temperatura = mes_usuario

    # Verificando o mês com menor temperatura
    if temperatura < menor_temperatura:
        menor_temperatura = temperatura
        mes_menor_temperatura = mes_usuario
```

### Calculando a temperatura média

```python
temperatura_media = sum(temperaturas) / len(temperaturas)
```

### Exibindo os resultados

```python

meses = ["janeiro", "fevereiro", "março", "abril", "maio", "junho", 
         "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]

print(f"\nTemperatura média máxima anual: {temperatura_media:.2f}°C")
print(f"Quantidade de meses escaldantes: {meses_escaldantes}")
print(f"Mês mais escaldante: {meses[mes_maior_temperatura - 1]} com {maior_temperatura}°C")
print(f"Mês menos quente: {meses[mes_menor_temperatura - 1]} com {menor_temperatura}°C")
```

Este programa pede para o usuário digitar a temperatura máxima de cada mês do ano.

Ele:
- Garante que o mês e a temperatura sejam válidos.
- Guarda todas as temperaturas em uma lista.
- Conta quantos meses tiveram temperaturas acima de 33°C (meses escaldantes).
- Mostra qual foi o mês mais quente e o menos quente.
- Calcula e exibe a média das temperaturas do ano.

No final, mostra um resumo com todas essas informações.
